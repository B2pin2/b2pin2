```python
%load_ext autoreload
%autoreload 2
import sys
sys.path.append('../')
import pandas as pd
```

```python
# ファイル名をここで指定する
file_name = 'XX.csv' # data/CSVsに入れたcsvファイル名を入力

# シャッフルしたい列名とランダムシードを指定
target_name = 'XXX'
random_seed = 9

# データフレームの作成
df = pd.read_csv(f'../data/CSVs/{file_name}') 
df # 念のため確認
```

```python
import random

target_li = df[target_name].values.tolist()

shuffled_li = target_li # 念のため新しいリストを用意
random.seed(random_seed)
random.shuffle(shuffled_li)

df[target_name] = shuffled_li
df # 念のため確認
```

```python
# 新しいcsvを保存（どの列をシャッフルしたのかは記憶しておく）
df.to_csv(f'../data/CSVs/{file_name[:-4]}_shuffled_seed{random_seed}.csv') 
```
